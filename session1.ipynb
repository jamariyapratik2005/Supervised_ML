## Question 2
**Given the following scenarios, classify each as supervised, unsupervised, or reinforcement learning:**
1. Netflix recommending movies based on your watch history
2. Spotify grouping similar songs into playlists
3. A self-driving car learning to park by trial and error

### Answer

**1) Netflix recommending movies based on your watch history → Supervised Learning**
> Netflix trains its model on labeled data consisting of user–movie interaction pairs (watched, rated, skipped), learning to predict a rating or engagement score for unseen movies for each user.

**2) Spotify grouping similar songs into playlists → Unsupervised Learning**
> Spotify uses clustering algorithms (e.g., K-Means) to group songs by audio features like tempo, energy, and genre without predefined labels, discovering natural groupings in the data.

**3) A self-driving car learning to park by trial and error → Reinforcement Learning**
> The car (agent) takes actions (steer, accelerate, brake) in an environment, receives rewards for getting closer to a correct parking position and penalties for collisions, and improves its parking policy over many attempts.

## Question 4
**Imagine you are building a spam filter for WhatsApp messages. Write a short paragraph explaining how traditional programming would approach this problem versus how machine learning would approach it.**

*Hint: Focus on rules vs learning from data.*

### Answer

**Traditional Programming Approach:**
In a traditional rule-based approach, a developer would manually write a long list of if-else rules to classify messages as spam. For example: *"If the message contains the word 'FREE' in all caps, or has more than 3 URLs, or comes from an unknown number with the word 'winner' — mark it as spam."* The programmer must anticipate every possible pattern spammers could use and hard-code each one. This is brittle because spammers constantly evolve their tactics (using "FR33" instead of "FREE," or embedding text in images), and each new trick requires a developer to manually update the rules. The system can never catch patterns the programmer hasn't explicitly thought of.

**Machine Learning Approach:**
With machine learning, instead of writing rules, we feed the system thousands of example WhatsApp messages that are already labeled as "spam" or "not spam." The ML algorithm (e.g., Naïve Bayes or a neural network) automatically discovers patterns — word frequencies, message length, presence of links, sender behavior, timing patterns — that distinguish spam from legitimate messages. When a new, unseen message arrives, the model predicts its class based on learned patterns. The key advantage is that the model generalizes: it can catch novel spam variants it has never seen before, and it improves over time as more labeled data is added — without a human needing to manually write new rules.

> **In short:** Traditional programming says *"I'll tell the computer exactly what spam looks like"*; machine learning says *"I'll show the computer examples and let it figure out what spam looks like on its own."*

## Question 7
**Given a simple linear regression model predicting song popularity from danceability, intentionally use only 5% of the data for training and 95% for testing. Observe the model's performance and explain whether this is likely to cause underfitting or overfitting.**

*Hint: Look at the accuracy or error on both train and test sets to support your answer.*

````python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Use only 'danceability' as the single feature
X_single = df[['danceability']]
y_single = df['popularity']

# Intentionally bad split: 5% train, 95% test
X_train_small, X_test_large, y_train_small, y_test_large = train_test_split(
    X_single, y_single, test_size=0.95, random_state=42
)

print(f"Training set size : {len(X_train_small)} rows  (5%)")
print(f"Testing set size  : {len(X_test_large)} rows (95%)")

# Train a simple Linear Regression model
model = LinearRegression()
model.fit(X_train_small, y_train_small)

# Predictions on both sets
y_pred_train = model.predict(X_train_small)
y_pred_test = model.predict(X_test_large)

# Performance metrics
train_rmse = np.sqrt(mean_squared_error(y_train_small, y_pred_train))
test_rmse  = np.sqrt(mean_squared_error(y_test_large, y_pred_test))
train_r2   = r2_score(y_train_small, y_pred_train)
test_r2    = r2_score(y_test_large, y_pred_test)

print(f"\n{'='*45}")
print(f"{'Metric':<20} {'Train (5%)':<15} {'Test (95%)'}")
print(f"{'='*45}")
print(f"{'RMSE':<20} {train_rmse:<15.4f} {test_rmse:.4f}")
print(f"{'R² Score':<20} {train_r2:<15.4f} {test_r2:.4f}")
print(f"{'='*45}")
````

````python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Training data + regression line
axes[0].scatter(X_train_small, y_train_small, alpha=0.3, s=10, color='dodgerblue')
axes[0].plot(X_train_small.sort_values('danceability'),
             model.predict(X_train_small.sort_values('danceability')),
             color='red', linewidth=2)
axes[0].set_title(f'Training Set (5% = {len(X_train_small)} rows)\nR² = {train_r2:.4f}', fontsize=12)
axes[0].set_xlabel('Danceability')
axes[0].set_ylabel('Popularity')

# Plot 2: Test data + regression line
axes[1].scatter(X_test_large, y_test_large, alpha=0.1, s=10, color='orange')
axes[1].plot(X_test_large.sort_values('danceability'),
             model.predict(X_test_large.sort_values('danceability')),
             color='red', linewidth=2)
axes[1].set_title(f'Test Set (95% = {len(X_test_large)} rows)\nR² = {test_r2:.4f}', fontsize=12)
axes[1].set_xlabel('Danceability')
axes[1].set_ylabel('Popularity')

plt.suptitle('Linear Regression: Danceability → Popularity (5/95 Split)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

### Explanation — Underfitting

This scenario most likely causes **underfitting**, for two combined reasons:

1. **Too little training data (only 5%):** The model sees a very small fraction of the data, so it cannot learn the full range of patterns and variations in the relationship between danceability and popularity. It has insufficient examples to generalize from.

2. **Too simple a model (single-feature linear regression):** A straight line using just one feature (`danceability`) is far too simplistic to capture the complex, multi-factor relationship that drives song popularity.

**Evidence:** Both the training R² and testing R² scores are very low (close to 0), meaning the model explains almost none of the variance in popularity on **either** set. This is the hallmark of underfitting — the model performs poorly on both training and test data, not just on test data.

> **Key Insight:** If it were overfitting, we would see **high training accuracy but low test accuracy**. Here, both are low → **underfitting**.

### Short Note: Bias–Variance Tradeoff in a Zomato Rating Predictor

Imagine building an ML model to predict a restaurant's rating on Zomato based on features like cuisine type, location, price range, delivery time, and number of reviews.

**High Bias (Underfitting):**
If we use a very simple model (e.g., predicting the average rating for all restaurants regardless of features), it will consistently predict ratings around 3.5–4.0 for every restaurant. It will miss important patterns — for example, that fine-dining restaurants in premium areas tend to have higher ratings. The model is too "rigid" and makes strong, incorrect assumptions.

**High Variance (Overfitting):**
If we use an overly complex model (e.g., a deep decision tree that memorizes every restaurant in the training data), it might perfectly predict ratings for known restaurants but fail badly on new ones. For instance, it might learn that "Restaurant X in Sector 18 with 247 reviews always gets 4.3" — but this memorization doesn't help predict ratings for a new restaurant it has never seen.

**The Sweet Spot:**
The ideal Zomato predictor finds the right balance — a model complex enough to learn genuine patterns (e.g., higher-priced restaurants with fast delivery tend to get better ratings) but simple enough to generalize to unseen restaurants. This is typically achieved through techniques like cross-validation, regularization, and careful feature selection.

## Question 9
**Give a real example (outside of healthcare/hospital) of overfitting from any app you use (e.g., Instagram, Flipkart, Swiggy), and explain in 2-3 lines why it might happen in that scenario.**

### Answer

**Example — Flipkart's Product Recommendation System:**

If you search for and buy a washing machine on Flipkart, the recommendation model might "overfit" to that purchase and keep showing you washing machines, washing machine accessories, and similar appliances for weeks — even though you've already bought one and are unlikely to buy another. This happens because the model memorizes your recent behavior (a single purchase event) too aggressively and treats it as a strong, ongoing preference, rather than recognizing that buying a washing machine is typically a **one-time event**. A well-generalized model would instead shift to recommending complementary products (detergent, clothes dryer) or move on to entirely different categories based on your broader browsing history.
