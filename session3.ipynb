## Question 2
**For the 'player_age' column in your IPL dataset, fill all missing values with the median age using pandas' `fillna()` method and display the updated column.**

```python
# Check missing values in 'player_age' BEFORE filling
print("BEFORE fillna()")
print("=" * 45)
print(f"Missing values in 'player_age': {df['player_age'].isnull().sum()}")
print(f"Median age: {df['player_age'].median()}")
print()

# Show the rows that have missing player_age
print("Rows with missing player_age:")
print(df[df['player_age'].isnull()][['player_name', 'player_age', 'team']])
print()
```

```python
# Fill missing values with MEDIAN age
median_age = df['player_age'].median()
df['player_age'] = df['player_age'].fillna(median_age)

# Display the updated column
print("AFTER fillna(median)")
print("=" * 45)
print(f"Missing values in 'player_age': {df['player_age'].isnull().sum()}")
print(f"Median used for filling: {median_age}")
print(f"\nUpdated 'player_age' column:")
print(df[['player_name', 'player_age']].to_string(index=True))
```

### Why use Median?
- **Median** is preferred over mean for filling missing values because it is **not affected by outliers**.
- For example, if one player's age were incorrectly entered as 99, the mean would be skewed, but the median would remain stable.

## Question 3
**Suppose the 'team' column in your IPL dataset contains missing values. Replace all missing team names with the constant value 'Unknown' and print the first 10 rows.**

```python
# Check missing values in 'team' BEFORE filling
print("BEFORE fillna()")
print("=" * 45)
print(f"Missing values in 'team': {df['team'].isnull().sum()}")
print("\nRows with missing team:")
print(df[df['team'].isnull()][['player_name', 'team']])
print()
```

```python
# Replace missing team names with 'Unknown'
df['team'] = df['team'].fillna('Unknown')

# Print the first 10 rows
print("AFTER fillna('Unknown') — First 10 Rows")
print("=" * 55)
print(f"Missing values in 'team': {df['team'].isnull().sum()}")
print()
print(df[['player_name', 'team']].head(10).to_string(index=True))
```

```python
# Verify: Show all rows that were filled with 'Unknown'
print("\nAll rows where team = 'Unknown':")
print(df[df['team'] == 'Unknown'][['player_name', 'team']].to_string(index=True))
```

### Why use a constant value?
- For **categorical columns** (like team names), we cannot use mean or median.
- Using a constant like `'Unknown'` preserves the data without introducing false information.
- Alternatively, we could use the **mode** (most frequent team), but that could introduce bias.

## Question 4
**Take the 'venue' column (categorical) from your IPL dataset and apply one-hot encoding using pandas' `get_dummies()`. Show the resulting DataFrame with the new columns.**

```python
# BEFORE encoding — show the venue column
print("BEFORE One-Hot Encoding")
print("=" * 40)
print(f"Unique venues: {df['venue'].nunique()}")
print(f"\nVenue values:")
print(df['venue'].value_counts())
```

```python
# Apply One-Hot Encoding using get_dummies()
df_encoded = pd.get_dummies(df, columns=['venue'], prefix='venue')

# Show the new columns created
venue_cols = [col for col in df_encoded.columns if col.startswith('venue_')]

print("AFTER One-Hot Encoding")
print("=" * 50)
print(f"Original columns: {len(df.columns)}")
print(f"New columns after encoding: {len(df_encoded.columns)}")
print(f"\nNew venue columns created ({len(venue_cols)}):")
for i, col in enumerate(venue_cols, 1):
    print(f"  {i}. {col}")
```

```python
# Display the resulting DataFrame (player_name + venue columns)
print("\nOne-Hot Encoded Venue Columns (first 10 rows):")
print("=" * 80)
display_cols = ['player_name'] + venue_cols
df_encoded[display_cols].head(10)
```

### How One-Hot Encoding Works

| Original `venue` | venue_Ahmedabad | venue_Chepauk | venue_Chinnaswamy | venue_Wankhede | ... |
|-------------------|:-:|:-:|:-:|:-:|:-:|
| Chinnaswamy | 0 | 0 | **1** | 0 | 0 |
| Wankhede | 0 | 0 | 0 | **1** | 0 |
| Chepauk | 0 | **1** | 0 | 0 | 0 |

- Each unique venue becomes a **new binary column** (0 or 1).
- Exactly **one column is 1** per row (the venue for that match).
- This allows ML models to work with categorical data that has **no ordinal relationship** (no venue is "greater" than another).

## Question 5
**Use `LabelEncoder` from sklearn to encode the 'player_role' column (e.g., Batsman, Bowler, Allrounder) in your IPL dataset, and display the mapping from original roles to encoded values.**

*Hint: Use LabelEncoder's `classes_` attribute to see the mapping.*

```python
# BEFORE encoding
print("BEFORE Label Encoding")
print("=" * 40)
print(f"Unique roles: {df['player_role'].nunique()}")
print(f"\nRole distribution:")
print(df['player_role'].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
le = LabelEncoder()

# Fit and transform the 'player_role' column
df['player_role_encoded'] = le.fit_transform(df['player_role'])

# Display the mapping using classes_ attribute
print("LABEL ENCODING MAPPING")
print("=" * 40)
print(f"\nMapping (classes_ attribute):")
for encoded_val, original_label in enumerate(le.classes_):
    print(f"  {original_label:<18s} → {encoded_val}")

print(f"\n{'='*60}")
print("Player Role — Before vs After Encoding")
print(f"{'='*60}")
print(df[['player_name', 'player_role', 'player_role_encoded']].to_string(index=True))
```

### Label Encoding vs One-Hot Encoding — When to Use Which?

| Method | When to Use | Example |
|--------|------------|----------|
| **Label Encoding** | When categories have an **ordinal (ranked) relationship** | Low → 0, Medium → 1, High → 2 |
| **One-Hot Encoding** | When categories have **no ordinal relationship** | Batsman, Bowler, Allrounder (no ranking) |

### ⚠️ Caution
- Label Encoding assigns numbers like 0, 1, 2, 3 — some ML models may incorrectly interpret this as **Allrounder (0) < Batsman (1) < Bowler (2)**, which has no real meaning.
- For **non-ordinal** categorical data, **One-Hot Encoding** is generally safer.
- However, **tree-based models** (Decision Trees, Random Forest, XGBoost) can handle Label Encoding well.
